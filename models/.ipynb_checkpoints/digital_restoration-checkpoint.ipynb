```python
import shutil

CNN_DATASET = "/content/cnn_dataset"

if os.path.exists(CNN_DATASET):
    shutil.rmtree(CNN_DATASET)

os.makedirs(
    CNN_DATASET + "/damaged"
)

os.makedirs(
    CNN_DATASET + "/not_damaged"
)

print("CNN dataset folders created.")
```

```python
image_lookup = {}

for path in image_files:

    filename = os.path.basename(path)

    image_lookup[filename] = path

print(
    "Indexed images:",
    len(image_lookup)
)
```

```python
def load_json(path):

    with open(path, "r") as f:
        return json.load(f)
```

```python
annotation_files = [
    lpl_train_json,
    lpl_test_json,
    stucco_train_json,
    stucco_test_json
]

annotations = {}

for path in annotation_files:

    if path is not None:

        annotations[
            os.path.basename(path)
        ] = load_json(path)

print(
    "Loaded annotation files:",
    len(annotations)
)
```

```python
for name, data in annotations.items():

    print("\n", name)

    if isinstance(data, dict):

        print(
            "Number of entries:",
            len(data)
        )

    break
```

```python
def get_regions(entry):

    if "regions" in entry:

        regions = entry["regions"]

        if isinstance(regions, dict):
            regions = list(regions.values())

        return regions

    return []
```

```python
def get_polygon(region):

    shape = region.get(
        "shape_attributes",
        {}
    )

    xs = shape.get(
        "all_points_x",
        []
    )

    ys = shape.get(
        "all_points_y",
        []
    )

    if len(xs) < 3 or len(ys) < 3:
        return None

    return list(
        zip(xs, ys)
    )
```

```python
damage_annotations = {}

for json_name, data in annotations.items():

    if not isinstance(data, dict):
        continue

    for key, entry in data.items():

        if not isinstance(entry, dict):
            continue

        filename = entry.get(
            "filename"
        )

        if filename is None:
            continue

        polygons = []

        for region in get_regions(entry):

            polygon = get_polygon(region)

            if polygon is not None:

                polygons.append(
                    polygon
                )

        if filename not in damage_annotations:

            damage_annotations[filename] = []

        damage_annotations[filename].extend(
            polygons
        )

print(
    "Images with annotations:",
    len(damage_annotations)
)
```

```python
total_polygons = sum(
    len(x)
    for x in damage_annotations.values()
)

print(
    "Total polygons:",
    total_polygons
)
```

```python
from PIL import Image
import random

PATCH_SIZE = 224

damaged_count = 0

for filename, polygons in damage_annotations.items():

    if filename not in image_lookup:
        continue

    image_path = image_lookup[filename]

    image = Image.open(
        image_path
    ).convert("RGB")

    width, height = image.size

    for polygon_id, polygon in enumerate(polygons):

        xs = [
            p[0]
            for p in polygon
        ]

        ys = [
            p[1]
            for p in polygon
        ]

        min_x = max(
            0,
            int(min(xs))
        )

        max_x = min(
            width,
            int(max(xs))
        )

        min_y = max(
            0,
            int(min(ys))
        )

        max_y = min(
            height,
            int(max(ys))
        )

        if max_x <= min_x or max_y <= min_y:
            continue

        crop = image.crop(
            (
                min_x,
                min_y,
                max_x,
                max_y
            )
        )

        # Ignore extremely tiny regions
        if crop.width < 20 or crop.height < 20:
            continue

        crop = crop.resize(
            (PATCH_SIZE, PATCH_SIZE)
        )

        output_path = (
            f"{CNN_DATASET}/damaged/"
            f"{Path(filename).stem}_"
            f"{polygon_id}.jpg"
        )

        crop.save(
            output_path,
            quality=95
        )

        damaged_count += 1

print(
    "Damaged patches:",
    damaged_count
)
```

```python
not_damaged_count = 0

random.seed(42)

for filename in damage_annotations.keys():

    if filename not in image_lookup:
        continue

    image_path = image_lookup[filename]

    image = Image.open(
        image_path
    ).convert("RGB")

    width, height = image.size

    polygons = damage_annotations[
        filename
    ]

    boxes = []

    for polygon in polygons:

        xs = [
            p[0]
            for p in polygon
        ]

        ys = [
            p[1]
            for p in polygon
        ]

        boxes.append(
            (
                min(xs),
                min(ys),
                max(xs),
                max(ys)
            )
        )

    attempts = 0
    generated = 0

    while (
        generated < 5
        and attempts < 100
    ):

        attempts += 1

        crop_width = min(
            224,
            width
        )

        crop_height = min(
            224,
            height
        )

        x = random.randint(
            0,
            max(0, width - crop_width)
        )

        y = random.randint(
            0,
            max(0, height - crop_height)
        )

        candidate = (
            x,
            y,
            x + crop_width,
            y + crop_height
        )

        # Check whether candidate overlaps
        # a damage bounding box

        overlap = False

        for box in boxes:

            bx1, by1, bx2, by2 = box

            cx1, cy1, cx2, cy2 = candidate

            if not (
                cx2 < bx1 or
                cx1 > bx2 or
                cy2 < by1 or
                cy1 > by2
            ):

                overlap = True
                break

        if overlap:
            continue

        crop = image.crop(
            candidate
        )

        crop = crop.resize(
            (PATCH_SIZE, PATCH_SIZE)
        )

        output_path = (
            f"{CNN_DATASET}/not_damaged/"
            f"{Path(filename).stem}_"
            f"normal_{generated}.jpg"
        )

        crop.save(
            output_path,
            quality=95
        )

        generated += 1
        not_damaged_count += 1


print(
    "Not-damaged patches:",
    not_damaged_count
)
```

```python
import tensorflow as tf

IMG_SIZE = (224, 224)
BATCH_SIZE = 32
SEED = 42

train_ds = tf.keras.utils.image_dataset_from_directory(

    CNN_DATASET,

    validation_split=0.20,

    subset="training",

    seed=SEED,

    image_size=IMG_SIZE,

    batch_size=BATCH_SIZE,

    label_mode="binary"
)
```

```python
validation_ds = tf.keras.utils.image_dataset_from_directory(

    CNN_DATASET,

    validation_split=0.20,

    subset="validation",

    seed=SEED,

    image_size=IMG_SIZE,

    batch_size=BATCH_SIZE,

    label_mode="binary"
)
```

```python
print(
    "Classes:",
    train_ds.class_names
)
```

```python
AUTOTUNE = tf.data.AUTOTUNE

train_ds = train_ds.cache().shuffle(
    1000
).prefetch(
    AUTOTUNE
)

validation_ds = validation_ds.cache().prefetch(
    AUTOTUNE
)
```

```python
model = tf.keras.Sequential([

    tf.keras.layers.Input(
        shape=(224, 224, 3)
    ),

    # Data augmentation
    tf.keras.layers.RandomFlip(
        "horizontal"
    ),

    tf.keras.layers.RandomRotation(
        0.05
    ),

    tf.keras.layers.RandomZoom(
        0.10
    ),

    # CNN Layer 1
    tf.keras.layers.Conv2D(
        32,
        (3, 3),
        activation="relu",
        padding="same"
    ),

    tf.keras.layers.MaxPooling2D(),

    # CNN Layer 2
    tf.keras.layers.Conv2D(
        64,
        (3, 3),
        activation="relu",
        padding="same"
    ),

    tf.keras.layers.MaxPooling2D(),

    # CNN Layer 3
    tf.keras.layers.Conv2D(
        128,
        (3, 3),
        activation="relu",
        padding="same"
    ),

    tf.keras.layers.MaxPooling2D(),

    # CNN Layer 4
    tf.keras.layers.Conv2D(
        256,
        (3, 3),
        activation="relu",
        padding="same"
    ),

    tf.keras.layers.MaxPooling2D(),

    # Classifier
    tf.keras.layers.GlobalAveragePooling2D(),

    tf.keras.layers.Dense(
        128,
        activation="relu"
    ),

    tf.keras.layers.Dropout(
        0.5
    ),

    # Binary output
    tf.keras.layers.Dense(
        1,
        activation="sigmoid"
    )
])

model.summary()
```

```python
model.compile(

    optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.001
    ),

    loss="binary_crossentropy",

    metrics=[
        "accuracy"
    ]
)
```

```python
history = model.fit(

    train_ds,

    validation_data=validation_ds,

    epochs=5
)
```

```python
plt.figure(figsize=(8, 5))

plt.plot(
    history.history["accuracy"],
    marker="o",
    label="Training Accuracy"
)

plt.plot(
    history.history["val_accuracy"],
    marker="o",
    label="Validation Accuracy"
)

plt.xlabel("Epoch")
plt.ylabel("Accuracy")

plt.title(
    "ArtInsight CNN - Damage Classification"
)

plt.legend()
plt.grid(True)

plt.show()

plt.figure(figsize=(8, 5))

plt.plot(
    history.history["loss"],
    marker="o",
    label="Training Loss"
)

plt.plot(
    history.history["val_loss"],
    marker="o",
    label="Validation Loss"
)

plt.xlabel("Epoch")
plt.ylabel("Loss")

plt.title(
    "ArtInsight CNN - Training Loss"
)

plt.legend()
plt.grid(True)

plt.show()
MODEL_PATH = "/content/artinsight_damage_cnn.keras"

model.save(
    MODEL_PATH
)

print(
    "Model saved to:",
    MODEL_PATH
)
from google.colab import files

files.download(
    MODEL_PATH
)
```

```python
def predict_damage(image_path):

    image = tf.keras.utils.load_img(
        image_path,
        target_size=IMG_SIZE
    )

    image_array = tf.keras.utils.img_to_array(
        image
    )

    image_array = image_array / 255.0

    image_array = np.expand_dims(
        image_array,
        axis=0
    )

    probability = model.predict(
        image_array,
        verbose=0
    )[0][0]

    class_names = ['damaged', 'not_damaged']

    if probability >= 0.5:

        predicted_class = class_names[1]

        confidence = probability

    else:

        predicted_class = class_names[0]

        confidence = 1 - probability

    return predicted_class, confidence
```

```python
test_image = "/content/test_image.jpg"

prediction, confidence = predict_damage(
    test_image
)

print(
    "Prediction:",
    prediction
)

print(
    f"Confidence: {confidence * 100:.2f}%"
)
```

```python
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt

IMG_SIZE = (224, 224)
BATCH_SIZE = 32
SEED = 42

DATASET_PATH = "/content/cnn_dataset"

# 70% training
train_ds = tf.keras.utils.image_dataset_from_directory(
    DATASET_PATH,
    validation_split=0.30,
    subset="training",
    seed=SEED,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    label_mode="binary"
)

# 30% temporary dataset
temp_ds = tf.keras.utils.image_dataset_from_directory(
    DATASET_PATH,
    validation_split=0.30,
    subset="validation",
    seed=SEED,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    label_mode="binary"
)

print("Classes:", train_ds.class_names)
```

```python
temp_batches = tf.data.experimental.cardinality(
    temp_ds
).numpy()

test_batches = temp_batches // 2

test_ds = temp_ds.take(test_batches)

validation_ds = temp_ds.skip(test_batches)

print(
    "Train batches:",
    tf.data.experimental.cardinality(train_ds).numpy()
)

print(
    "Validation batches:",
    tf.data.experimental.cardinality(validation_ds).numpy()
)

print(
    "Test batches:",
    tf.data.experimental.cardinality(test_ds).numpy()
)
```

```python
from tensorflow.keras.callbacks import (
    EarlyStopping,
    ModelCheckpoint,
    ReduceLROnPlateau
)

early_stopping = EarlyStopping(
    monitor="val_loss",
    patience=5,
    restore_best_weights=True,
    verbose=1
)

checkpoint = ModelCheckpoint(
    "/content/best_artinsight_cnn.keras",
    monitor="val_accuracy",
    save_best_only=True,
    verbose=1
)

reduce_lr = ReduceLROnPlateau(
    monitor="val_loss",
    factor=0.5,
    patience=2,
    min_lr=1e-6,
    verbose=1
)
```

```python
history = model.fit(
    train_ds,

    validation_data=validation_ds,

    epochs=10,

    callbacks=[
        early_stopping,
        checkpoint,
        reduce_lr
    ]
)
```

```python
model = tf.keras.models.load_model(
    "/content/best_artinsight_cnn.keras"
)

print("Best model loaded successfully.")
test_loss, test_accuracy = model.evaluate(
    test_ds,
    verbose=1
)

print("\n==============================")
print("FINAL TEST RESULTS")
print("==============================")

print(
    f"Test Loss: {test_loss:.4f}"
)

print(
    f"Test Accuracy: {test_accuracy * 100:.2f}%"
)
from sklearn.metrics import (
    classification_report,
    confusion_matrix
)

y_true = []
y_pred = []

for images, labels in test_ds:

    predictions = model.predict(
        images,
        verbose=0
    )

    predictions = (
        predictions.ravel() >= 0.5
    ).astype(int)

    y_true.extend(
        labels.numpy().astype(int)
    )

    y_pred.extend(
        predictions
    )

y_true = np.array(y_true)
y_pred = np.array(y_pred)
class_names = train_ds.class_names

print(
    classification_report(
        y_true,
        y_pred,
        target_names=class_names,
        digits=4
    )
)
cm = confusion_matrix(
    y_true,
    y_pred
)

print(cm)
import seaborn as sns

plt.figure(
    figsize=(7, 6)
)

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    xticklabels=class_names,
    yticklabels=class_names
)

plt.xlabel("Predicted")
plt.ylabel("Actual")

plt.title(
    "CNN Damage Classification - Confusion Matrix"
)

plt.show()
plt.figure(figsize=(8, 5))

plt.plot(
    history.history["accuracy"],
    marker="o",
    label="Training Accuracy"
)

plt.plot(
    history.history["val_accuracy"],
    marker="o",
    label="Validation Accuracy"
)

plt.xlabel("Epoch")
plt.ylabel("Accuracy")

plt.title(
    "Training and Validation Accuracy"
)

plt.legend()

plt.show()
plt.figure(figsize=(8, 5))

plt.plot(
    history.history["loss"],
    marker="o",
    label="Training Loss"
)

plt.plot(
    history.history["val_loss"],
    marker="o",
    label="Validation Loss"
)

plt.xlabel("Epoch")
plt.ylabel("Loss")

plt.title(
    "Training and Validation Loss"
)

plt.legend()

plt.show()

FINAL_MODEL_PATH = (
    "/content/artinsight_damage_cnn_final.keras"
)

model.save(
    FINAL_MODEL_PATH
)

print(
    "Model saved:",
    FINAL_MODEL_PATH
)
```
